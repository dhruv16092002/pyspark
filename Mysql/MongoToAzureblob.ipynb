```python
from urllib.parse import quote_plus

dbutils.widgets.text("source host", "", "Source host")
dbutils.widgets.text("source port", "", "Source port")
dbutils.widgets.text("source database name", "", "Source Database Name")
dbutils.widgets.text("source user name", "", "Source Username")
dbutils.widgets.text("source password", "", "Source password")
dbutils.widgets.text("source collection name", "", "Source table name")

source_password = quote_plus(dbutils.widgets.get("source password"))
source_host = dbutils.widgets.get("source host")
source_port = dbutils.widgets.get("source port")
source_user_name = dbutils.widgets.get("source user name")
source_database_name = dbutils.widgets.get("source database name")
source_collection_name = dbutils.widgets.get("source collection name")
source_url = f"mongodb://{source_user_name}:{source_password}@{source_host}:{source_port}/{source_database_name}.employee?authSource=admin"
```

```python
dbutils.widgets.text("sink account name", "", "Sink account name")
dbutils.widgets.text("sink container name", "", "Sink container name")
dbutils.widgets.text("sink account key", "", "Sink account key")
dbutils.widgets.text("sink file path", "", "Sink file path")

sink_container_name = dbutils.widgets.get("sink container name")
sink_account_name = dbutils.widgets.get("sink account name")
sink_destination_file_path = dbutils.widgets.get("sink file path")
sink_account_key = dbutils.widgets.get("sink account key")

spark.conf.set(f"fs.azure.account.key.{sink_account_name}.blob.core.windows.net", sink_account_key)
sink_url = f"wasbs://{sink_container_name}@{sink_account_name}.blob.core.windows.net/{sink_destination_file_path}"
```

```python
df = spark.read.format('com.mongodb.spark.sql.DefaultSource').option( "uri", source_url).load()
```

```python
df.write.format("parquet").mode("overwrite").save(sink_url)
```
