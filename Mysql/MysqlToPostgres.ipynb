```python
dbutils.widgets.text("soure host", "", "Source host")
dbutils.widgets.text("soure port", "", "Source port")
dbutils.widgets.text("soure database_name", "", "Source Database Name")
dbutils.widgets.text("soure user_name", "", "Source Username")
dbutils.widgets.text("soure password", "", "Source password")
dbutils.widgets.text("soure table", "", "Source table name")

dbutils.widgets.text("sink host", "", "Sink host")
dbutils.widgets.text("sink port", "", "Sink port")
dbutils.widgets.text("sink database_name", "", "Sink Database Name")
dbutils.widgets.text("sink user_name", "", "Sink Username")
dbutils.widgets.text("sink password", "", "Sink password")
dbutils.widgets.text("sink table", "", "Sink table name")

```

```python
soure_type = "mysql"
soure_host = dbutils.widgets.get("soure host")
soure_port = int(dbutils.widgets.get("soure port"))
soure_database_name = dbutils.widgets.get("soure database_name")
soure_user_name = dbutils.widgets.get("soure user_name")
soure_password = dbutils.widgets.get("soure password")
soure_table_name = dbutils.widgets.get("soure table")
soure_url =  f"jdbc:{soure_type}://{soure_host}:{soure_port}/{soure_database_name}"
```

```python
sink_type = "postgresql"
sink_host = dbutils.widgets.get("sink host")
sink_port = int(dbutils.widgets.get("sink port"))
sink_database_name = dbutils.widgets.get("sink database_name")
sink_user_name = dbutils.widgets.get("sink user_name")
sink_password = dbutils.widgets.get("sink password")
sink_table_name = dbutils.widgets.get("sink table")
sink_url =  f"jdbc:{sink_type}://{sink_host}:{sink_port}/{sink_database_name}"
```

```python
df = spark.read.format("jdbc").options(
    url = soure_url,
    dbtable = soure_table_name,
    user = soure_user_name,
    password = soure_password
).load()
```

```python
"""Between We can transfrom the data"""
```

```python
database_table = f"{sink_database_name}.{sink_table_name}"

properties = {
    "user": sink_user_name,
    "password": sink_password,
    "driver": 'org.postgresql.Driver'
}
df.write.jdbc(url=sink_url, table=database_table, mode="overwrite", properties=properties)
```
